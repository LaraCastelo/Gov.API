```python
url = "https://api.portaldatransparencia.gov.br/api-de-dados/cartoes"
API_KEY = os.environ["API_KEY"]
headers = {"chave-api-dados": API_KEY}
params = {
    "mesExtratoInicio": "07/2025",
    "mesExtratoFim": "06/2026", #12 meses
    "tipoCartao": 1,  # CPGF = 1 (Cartão de Pagamento do Governo Federal)
    "pagina": 1
}

def buscar_dados():
    dados_pags = []

    while True:
        try:
            response = requests.get(url, headers=headers, params=params)
            response.raise_for_status()
            dados_json = response.json()

            if not dados_json:
                print("Fim real da paginação (resposta vazia).")
                break

            dados_pags.extend(dados_json)
            print(f"Página {params['pagina']}: {len(dados_pags)} registros no total")
            if params["pagina"] % 50 == 0:
                with open("dados_cartoes_progresso.json", "w", encoding="utf-8") as f:
                    json.dump(dados_pags, f, ensure_ascii=False)

            params["pagina"] += 1
            time.sleep(1.2)

        except requests.exceptions.RequestException as e:
            print(f"Erro na página {params['pagina']}: {e}")
            break  # sai do loop mas preserva o que já foi coletado

    return dados_pags

dados_pags = buscar_dados()
```
